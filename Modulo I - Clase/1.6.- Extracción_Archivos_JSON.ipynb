# Archivos JSON

JSON representa datos mediante pares `clave: valor`, listas y objetos anidados. Es muy habitual en APIs, configuraciones y registros de aplicaciones porque conserva mejor la estructura que una tabla plana.

El proceso tiene dos etapas: `json.load` o `json.loads` convierte el contenido a diccionarios y listas de Python; después, `json_normalize` ayuda a convertir objetos repetidos o anidados en columnas de un DataFrame. No todos los JSON tienen la misma forma, por lo que primero conviene inspeccionar sus claves y niveles.


El formato JSON (JavaScript Object Notation) es ampliamente utilizado para el intercambio de datos, especialmente en aplicaciones web y APIs. Python incluye la librería estándar `json` para leer y manipular archivos JSON.

#### Anatomía de un archivo JSON

JSON se construye con solo dos estructuras que se combinan y anidan, lo que le permite representar datos muy complejos:

- **Objetos** `{ }` → pares `clave: valor` (equivalen a un `dict` de Python).
- **Arreglos** `[ ]` → listas ordenadas de valores (equivalen a una `list`).

| JSON | Python |
|---|---|
| `{ }` *object* | `dict` |
| `[ ]` *array* | `list` |
| `"texto"` | `str` |
| `12`, `3.14` | `int`, `float` |
| `true` / `false` | `True` / `False` |
| `null` | `None` |

El archivo `data.json` es un **catálogo de productos** real: cada elemento tiene nombre, precio (`mrp`), código de barras y una **lista anidada** de `categories`.

````python
# Configuración común para ejecutar esta sección de forma independiente
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dir_base = os.getcwd()
print(dir_base)
ruta = os.path.join(dir_base, 'Data')
print(ruta)
````

````python
import json

json_data = '{"personas": [{"nombre": "Ana", "edad": 23}, {"nombre": "Luis", "edad": 31}]}'
data = json.loads(json_data)
print('Personas extraídas del archivo JSON:')
for persona in data['personas']:
    print(f'Nombre: {persona["nombre"]}, Edad: {persona["edad"]}')
````

````python
#json normalize, cargar un archivo json 
from pandas import json_normalize

with open(os.path.join(ruta, 'data.json')) as jsonfile:
    jsondata = json.load(jsonfile)
jsondata
````

````python
df_json = json_normalize(jsondata['data'])
df_json
````

````python
# Inspeccionar la estructura antes de normalizar (paso clave con JSON desconocido)
df_json['categories'][0]
````

#### Aplanar estructuras anidadas con `record_path` y `meta`

La columna `categories` es una **lista de objetos** dentro de cada producto. Con `record_path` le indicamos a `json_normalize` cuál es la lista que queremos "explotar" (una fila por categoría), y con `meta` qué campos del nivel superior conservar en cada fila.

````python
# Explotar la lista anidada 'categories' conservando datos del producto
df_cat = json_normalize(
    jsondata['data'],
    record_path='categories',
    meta=['itemName', 'barcode', 'mrp']
)
df_cat
````

#### Un mini análisis sobre el catálogo

Una vez que el JSON es un `DataFrame`, se analiza como cualquier tabla: estadísticas, ordenamientos y filtros.

````python
# Estadística y ordenamiento sobre los datos ya tabulares
df_cat['mrp'].mean()
````

````python
df_json[['itemName', 'mrp']].sort_values('mrp', ascending=False)
````

#### Del DataFrame de vuelta a JSON (serialización)

Extraer no es el único camino: también se **serializa** (guardar de Python a JSON). `json.dumps` convierte un objeto de Python a texto JSON, y `DataFrame.to_json` exporta una tabla completa. El argumento `indent` produce una salida legible y `ensure_ascii=False` / `force_ascii=False` conserva los acentos.

````python
# json.dumps: de un dict/list de Python a texto JSON con formato legible
resumen = {'tota_productos': len(jsondata['data']), 'productos': [p['itemName'] for p in jsondata['data']]}
resumen
````

````python
print(json.dumps(resumen, indent=2, ensure_ascii=True))
````

````python
# DataFrame.to_json exporta una tabla; orient='records' = lista de objetos
df_json
````

````python
df_json[['itemName', 'mrp']].to_json(orient='records', indent=2, force_ascii=True)
````

#### Atajo: `pd.read_json`

Cuando el JSON ya es una **lista de objetos "plana"** (sin anidamiento), `pd.read_json` crea el `DataFrame` en un solo paso. Es el equivalente de `pd.read_csv`, pero para JSON.

````python
# pd.read_json convierte una lista de objetos JSON en un DataFrame directamente
json_plano = '[{"ciudad": "Guadalajara", "temp": 28}, {"ciudad": "Monterrey", "temp": 34}]'
pd.read_json(json_plano)
````

## Práctica de Laboratorio: Extracción de archivos JSON

Utilizando el fichero metallica.json de de los discos del grupo Metallica, crea distintas funciones en python, utilizando la librería json, que realicen las siguientes funciones:

1. **Listar información:** Listar el título del disco, y el contenido y fecha en el que se ha publicado en la wiki.
2. **Contar información:** Mostrar los títulos de discos y el número de canciones que tiene cada una.
3. **Buscar o filtrar información:** Recibe un límite inferior y otro superior, y muestra el título, y los nombres de las etiquetas ("tags") de los discos cuyos número de oyentes esté entre los dos límites.
4. **Buscar información relacionada:** Recibe el nombre una canción y muestra el título del disco donde está.

````python
json_metalica_path = os.path.join(ruta, "metallica.json")
json_metalica_path
````

````python
with open(json_metalica_path, encoding='utf-8') as json_metalica:
    jsondata_metalica = json.load(json_metalica)
jsondata_metalica
````

````python
df_json_metalica = json_normalize(jsondata_metalica)
df_json_metalica
````

````python
# Listar información: Listar el título del disco, y el contenido y fecha en el que se ha publicado en la wiki.
df_json_metalica_info = df_json_metalica[['album.name', 'album.wiki.published', 'album.wiki.content']]
df_json_metalica_info
````

````python
# Contar información: Mostrar los títulos de discos y el número de canciones que tiene cada una.
canciones_albumnes = []
for tracks in df_json_metalica['album.tracks.track']:
    canciones_albumnes.append(len(tracks))
canciones_albumnes
        
````

````python
df_json_metalica_info.loc[:,'num_canciones'] = canciones_albumnes
df_json_metalica_info
````

````python
# Buscar o filtrar información: Recibe un límite inferior y otro superior, y muestra el título, y los nombres de las etiquetas ("tags") de los discos cuyos número de oyentes esté entre los dos límites.
df_json_metalica
````

````python
lim_inferior = 1000
lim_superior = 500000

tags_tag = []
for tags in df_json_metalica['album.tags.tag']:
    list_tag = []
    for tg in tags:
        list_tag.append(tg['name'])
    tags_tag.append(list_tag)
tags_tag
        
````

````python
df_json_metalica_info.loc[:,'tags_tag'] = tags_tag
df_json_metalica_info
````

````python
lim_inferior, lim_superior
````
