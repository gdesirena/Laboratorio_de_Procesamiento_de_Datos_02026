# Archivos de texto

Los archivos de texto no siempre tienen una estructura explícita. Para extraerlos hay que determinar si sus campos están separados por tabuladores, comas, espacios u otro delimitador, o si cada columna ocupa un ancho fijo.

### De texto a `DataFrame`

La estrategia elegida debe corresponder al formato real. Si el separador cambia entre filas, hay que normalizar la fuente antes de convertirla en tabla; si el archivo contiene texto libre, conviene leerlo como una cadena y aplicar técnicas de procesamiento de lenguaje.

### Formas de acceder a archivos (modos de apertura)

La función `open(archivo, modo)` recibe un **modo** que define qué operaciones se permiten sobre el archivo:

| Modo | Nombre | Descripción |
|---|---|---|
| `'r'` | read | Solo lectura (por defecto). Da error si el archivo no existe. |
| `'w'` | write | Escritura. **Crea** el archivo o **sobrescribe** su contenido. |
| `'a'` | append | Escritura al final; conserva el contenido existente. |
| `'x'` | exclusive | Crea un archivo nuevo; da error si ya existe. |
| `'r+'` | read/write | Lectura y escritura sin borrar el contenido previo. |
| `'b'` | binary | Modo binario (imágenes, parquet…); se combina, p. ej. `'rb'`. |
| `'t'` | text | Modo texto (por defecto); se combina, p. ej. `'rt'`. |

Conviene indicar la codificación con `encoding='utf-8'` para leer correctamente acentos y caracteres especiales. Los métodos de lectura más comunes son `read()` (todo el contenido como una cadena), `readline()` (una sola línea) y `readlines()` (una lista con todas las líneas).

```python
# Modo 'a': agrega contenido al final sin borrar lo existente
with open(ruta + '/inventario.txt', 'a', encoding='utf-8') as f:
    f.write('naranja,8\n')

with open(ruta + '/inventario.txt', 'r', encoding='utf-8') as f:
    print(f.read())
```

### Expresiones regulares

Las **expresiones regulares** (*regex*) describen patrones de texto. Son apropiadas cuando la fuente no tiene una estructura tabular, pero sí reglas repetibles: correos, fechas, identificadores o códigos.

El flujo habitual es: definir el patrón, probarlo con ejemplos representativos, extraer coincidencias y validar los casos que no coincidan. Una regex demasiado general puede capturar datos incorrectos; una demasiado estricta puede omitir variaciones válidas.


Las **expresiones regulares** (regex) son patrones que se utilizan para buscar, extraer o manipular texto de manera flexible y eficiente. Permiten identificar cadenas de texto que cumplen ciertas reglas, como correos electrónicos, números de teléfono, palabras específicas, etc.

##### ¿Cómo se usan en Python?

En Python, se utiliza el módulo `re` para trabajar con expresiones regulares. Algunas funciones comunes son:

- `re.search()`: busca un patrón en cualquier posición y devuelve la primera coincidencia.
- `re.match()`: comprueba la coincidencia desde el inicio de la cadena.
- `re.findall()`: devuelve todas las coincidencias como una lista.
- `re.split()`: divide el texto usando el patrón como separador.
- `re.sub()`: reemplaza coincidencias por otro texto.

##### ¿Cómo definir patrones en regex?

- Los patrones se definen como cadenas de texto; el prefijo `r` evita que Python interprete algunas barras invertidas antes que regex.
- Algunos caracteres especiales:
  - `.`: cualquier carácter excepto salto de línea
  - `\d`: un dígito (0-9)
  - `\w`: un carácter alfanumérico o guion bajo
  - `\s`: un espacio en blanco
  - `*`: cero o más repeticiones
  - `+`: una o más repeticiones
  - `?`: cero o una repetición
  - `^`: inicio de línea
  - `$`: fin de línea
  - `[abc]`: cualquier carácter a, b o c
  - `( )`: agrupación y captura

> **Lectura del ejemplo:** `\w+@\w+\.\w+` busca una estructura simple de correo. Es útil para aprender, pero no pretende validar todas las direcciones permitidas por los estándares de correo.


##### Ejemplo: Correos electrónicos

```python
import re

texto_correo = "Mi correo es gdesirena@iteso.mx pero anteriormente usaba gdesirena@gmail.com"
patron = r"\w+@\w+\.\w+"  # patrón para un correo electrónico

resultado = re.search(patron, texto_correo)
if resultado:
    print("Correo encontrado:", resultado.group())
```

```python
 #Encuentra todos los correos en texto_correo
re.findall(patron, texto_correo)
```

```python
#captura el dominio después del @
dominios = re.findall(r'@([\w\.-]+)', texto_correo)
```

##### Más ejemplos de expresiones regulares

Los siguientes ejemplos muestran otras funciones y patrones del módulo `re`: búsqueda de números de teléfono, extracción de precios, sustitución de texto (`re.sub`), la diferencia entre `re.match` y `re.search`, y la captura de **grupos**.

```python
# Búsqueda de números de teléfono con distintos separadores
texto_contacto = 'Teléfonos de contacto: 33-1234-5678, 55 8765 4321 y 3312345678.'
telefonos = re.findall(r'\d{2}[\s-]?\d{4}[\s-]?\d{4}', texto_contacto)
print('Teléfonos encontrados:', telefonos)
```

```python
# Extraer precios; el grupo entre paréntesis captura solo el número
texto_precios = 'Los precios son $1,250.00, $99.50 y $12,000.99 pesos.'
precios = re.findall(r'\$([\d,]+\.\d{2})', texto_precios)
print('Precios:', precios)
```

```python
# re.sub reemplaza coincidencias: aquí ocultamos los dígitos con '#'
oculto = re.sub(r'\d', '#', texto_contacto)
print(oculto)
```

```python
# re.match solo coincide al inicio de la cadena; re.search busca en cualquier posición
print('match "Tel"     :', re.match(r'Tel', texto_contacto))       # coincide
print('match "contacto":', re.match(r'contacto', texto_contacto))  # None (no está al inicio)
print('search "contacto":', re.search(r'contacto', texto_contacto))  # sí lo encuentra
```

```python
# Grupos con paréntesis para separar día, mes y año de una fecha
texto_fecha = 'La entrega fue el 15/09/2023.'
m = re.search(r'(\d{2})/(\d{2})/(\d{4})', texto_fecha)
if m:
    print('Fecha completa:', m.group(0))
    print('Día:', m.group(1), '| Mes:', m.group(2), '| Año:', m.group(3))
```

```python
# Grupos con nombre para extraer varios registros y pasarlos a un DataFrame
registros = 'Ana:23; Luis:31; Marta:45'
coincidencias = re.findall(r'(?P<nombre>\w+):(?P<edad>\d+)', registros)
pd.DataFrame(coincidencias, columns=['nombre', 'edad'])
```
