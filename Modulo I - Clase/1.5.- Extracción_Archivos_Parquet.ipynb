# Archivos Parquet

**Parquet** es un formato de almacenamiento **columnar** y **binario** diseñado para el análisis de datos a gran escala. A diferencia de un CSV (que guarda los datos por filas y en texto plano), Parquet organiza la información por columnas, lo que ofrece varias ventajas:

- **Compresión eficiente:** al agrupar datos del mismo tipo en cada columna, se comprime mucho mejor y ocupa menos espacio en disco.
- **Lectura selectiva:** permite leer solo las columnas necesarias sin cargar todo el archivo, acelerando las consultas.
- **Conserva los tipos de dato:** guarda el *esquema* (el tipo de cada columna), a diferencia del CSV donde todo es texto.
- **Estándar en Big Data:** es el formato preferido en herramientas como Apache Spark, Hadoop, AWS Athena y en los *data lakes*.

**¿Cuándo se usa?** Al trabajar con grandes volúmenes de datos, en flujos de trabajo analíticos y cuando se busca eficiencia en almacenamiento y velocidad de lectura.

En `pandas` se usa `df.to_parquet()` para escribir y `pd.read_parquet()` para leer. Requiere un motor instalado: `pyarrow` (recomendado) o `fastparquet`.

```python
# Creamos un DataFrame de ejemplo y lo guardamos en formato Parquet
df_parquet = pd.DataFrame({
    'producto': ['manzana', 'pera', 'plátano', 'naranja'],
    'precio': [3.5, 5.0, 2.0, 7.5],
    'cantidad': [10, 8, 20, 5]
})
#Guardando en formato parquet
```

```python
df_parquet.to_parquet(ruta+'/productos.parquet')
```

```python
# Lectura completa del archivo Parquet
pd.read_parquet(ruta+'/productos.parquet')
```

```python
# Ventaja de Parquet: leer únicamente algunas columnas
pd.read_parquet(ruta+'/productos.parquet', columns=['precio', 'cantidad'])
```

```python
df_p_leidos = pd.read_parquet(ruta+'/productos.parquet')
df_p_leidos
```



|  | Archivo CSV (Comma-Separated Values) | Archivo Apache Parquet |
| :--- | :--- | :--- |
| **Representación** | Un documento de **texto plano** basado en filas, diseñado para ser leído tanto por humanos como por máquinas. | El estándar de la industria para **Big Data**, un formato binario basado en columnas optimizado para analítica masiva. |
| **Orientación de datos** | **Por filas**. Lee toda la fila completa (registro por registro), incluso si solo necesitas una columna. | **Por columnas**. Lee solo las columnas específicas que necesitas, ignorando el resto. |
| **Legibilidad** | **Alta**. Se puede abrir en el Bloc de notas o Excel sin herramientas especiales. | **Baja**. Al ser un formato binario, requiere software de programación o herramientas de datos para abrirse. |
| **Almacenamiento** | **Inadecuado para archivos grandes**. No tiene compresión nativa y ocupa mucho espacio en disco. | **Altamente eficiente**. Utiliza algoritmos avanzados que reducen drásticamente el espacio en disco. |
| **Metadatos y Tipos** | **No incluye metadatos**. Todo se guarda como texto; el sistema debe adivinar si un dato es número, fecha o texto. | **Esquema estricto**. Almacena de forma nativa los tipos de datos (int, float, boolean, timestamps), evitando errores de interpretación. |

```python
dfs = []
for archivo in os.listdir(carpeta_csv):
    df = pd.read_csv(os.path.join(carpeta_csv,archivo))
    df['origen'] = os.path.basename(archivo)
    dfs.append(df)
consolidad_df = pd.concat(dfs, axis=0, ignore_index=True)
consolidad_df
```
