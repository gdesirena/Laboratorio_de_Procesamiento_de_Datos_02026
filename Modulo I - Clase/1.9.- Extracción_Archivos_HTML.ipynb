# Archivos HTML y web scraping

**HTML (HyperText Markup Language, o lenguaje de marcado de hipertexto)** es el lenguaje que describe la estructura de una página web. No es un lenguaje de programación: organiza el contenido mediante **elementos**, **etiquetas** y **atributos**. El estilo visual suele definirse con CSS y el comportamiento interactivo con JavaScript.

Algunas etiquetas frecuentes son:

| Etiqueta | Función | Ejemplo de información que puede contener |
|---|---|---|
| `html` | Elemento raíz del documento | Toda la página |
| `head` | Metadatos y recursos | título, codificación, hojas de estilo |
| `title` | Título de la página | nombre mostrado en la pestaña |
| `body` | Contenido visible | textos, tablas, imágenes y enlaces |
| `h1`, `h2`, `h3` | Encabezados de distintos niveles | títulos y subtítulos |
| `p` | Párrafo | descripción o resumen |
| `a` | Enlace | URL en el atributo `href` |
| `table`, `tr`, `th`, `td` | Tabla, fila, encabezado y celda | datos tabulares |
| `ul`, `ol`, `li` | Lista no ordenada, ordenada y elemento | categorías o pasos |
| `div`, `section`, `article` | Contenedores semánticos o estructurales | bloques de contenido |

Las etiquetas normalmente tienen apertura y cierre, por ejemplo, `<p>Texto</p>`. Los atributos aportan información adicional: `<a href="https://ejemplo.org">enlace</a>` contiene el atributo `href`.

### DOM y extracción

El navegador interpreta el HTML como un árbol llamado **DOM (Document Object Model, o modelo de objetos del documento)**. Cada etiqueta es un nodo que puede contener otros nodos. `BeautifulSoup` permite recorrer ese árbol, buscar elementos y obtener texto o atributos.

El **web scraping** es la extracción automatizada de información publicada en páginas web. Un flujo de trabajo responsable separa estas etapas:

1. Obtener el HTML desde un archivo local o una URL.
2. Verificar que la respuesta sea válida y analizarla con un parser.
3. Localizar los elementos mediante etiquetas, clases, identificadores o selectores CSS.
4. Limpiar y estructurar los resultados en listas, diccionarios o un `DataFrame`.
5. Validar los datos extraídos y guardar el resultado con una codificación adecuada.

Una página puede tener contenido estático, que aparece directamente en el HTML descargado, o contenido dinámico, que se genera después con JavaScript. `requests` y `BeautifulSoup` funcionan muy bien para HTML estático; para contenido dinámico puede ser necesario localizar una API pública o utilizar un navegador automatizado, siempre respetando las reglas del sitio.

## Herramientas y criterios de selección

| Herramienta | Uso principal | Cuándo elegirla |
|---|---|---|
| `open()` | Leer un HTML local | Cuando el archivo ya está descargado |
| `requests` | Descargar HTML desde una URL | Cuando el contenido está disponible en la respuesta HTTP |
| `BeautifulSoup` | Navegar y extraer elementos | Cuando se necesita controlar etiquetas, atributos y selectores |
| `pandas.read_html()` | Convertir tablas HTML | Cuando los datos están organizados en etiquetas `table` |
| `urllib.parse.urljoin()` | Convertir enlaces relativos en absolutos | Cuando una página usa rutas como `/noticias/1` |
| Selenium o Playwright | Automatizar un navegador | Cuando el contenido aparece después de ejecutar JavaScript |

## Analizar el árbol HTML y extraer elementos



`BeautifulSoup` ofrece varios caminos para localizar información. `find()` devuelve el primer elemento que coincide; `find_all()` devuelve todos; `select_one()` y `select()` aceptan selectores CSS.



El texto se obtiene con `.get_text(' ', strip=True)`. Los atributos se consultan como un diccionario, por ejemplo `element.get('href')`, que es más seguro que `element['href']` cuando el atributo podría no existir.

```python
# Extraer texto de los enlaces y sus atributos
enlaces = []
for enlace in soup.find_all('a', href=True):
    enlaces.append({
        'texto': enlace.get_text(' ', strip=True),
        'url': enlace.get('href'),
        'titulo': enlace.get('title'),
    })
df_enlaces = pd.DataFrame(enlaces)
df_enlaces.head()
```

```python
# Descargar HTML desde una URL y verificar la respuesta
import requests
from bs4 import BeautifulSoup
url = 'https://www.python.org/'
respuesta = requests.get(url)
respuesta.raise_for_status()

print('Código de estado:', respuesta.status_code)
print('Tipo de contenido:', respuesta.headers.get('content-type'))
soup_web = BeautifulSoup(respuesta.text, 'html.parser')
```

## Enlaces relativos y absolutos



Una página puede contener `https://sitio.org/articulo`, `/articulo` o `articulo`. Para guardar enlaces utilizables fuera de la página, conviene convertirlos a URL absolutas con `urljoin()`.

```python
from urllib.parse import urljoin
base_url = respuesta.url
df_enlaces_web = pd.DataFrame([
    {
        'texto': enlace.get_text(' ', strip=True),
        'url': urljoin(base_url, enlace.get('href')),
    }
    for enlace in soup_web.find_all('a', href=True)
])

df_enlaces_web.head()
```

###### Ejemplo sencillo webscraping 

[Link_products](https://codedamn-classrooms.github.io/webscraper-python-codedamn-classroom-website/)

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
 
url = "https://codedamn-classrooms.github.io/webscraper-python-codedamn-classroom-website/"
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
type(soup)
```

```python
# Obtener un DataFrame con la información de los productos
productos = []
for etiqueta_precio in soup.find_all(
    'h4',
    string=lambda texto: texto and texto.strip().startswith('$'),
):
    # El producto se encuentra en el siguiente enlace relacionado.
    enlace_producto = etiqueta_precio.find_next('a')
    if not enlace_producto:
        continue
    nombre_producto = enlace_producto.get_text(' ', strip=True)
    url_producto = enlace_producto.get('href')

    # La descripción se encuentra en el siguiente párrafo.
    etiqueta_descripcion = enlace_producto.find_next('p')
    descripcion = (
        etiqueta_descripcion.get_text(' ', strip=True)
        if etiqueta_descripcion else ''
    )
    # Las reseñas aparecen cerca del producto.
    etiqueta_resenas = enlace_producto.find_next(
        string=lambda texto: texto and 'review' in texto.lower()
    )

    try:
        resenas = int(etiqueta_resenas.strip().split()[0])
    except (AttributeError, ValueError, IndexError):
        resenas = None
    try:
        precio = float(etiqueta_precio.get_text(strip=True).replace('$', ''))
    except (ValueError, AttributeError):
        precio = None

    productos.append({
        'nombre_producto': nombre_producto,
        'precio': precio,
        'descripcion': descripcion,
        'resenas': resenas,
        'url_producto': url_producto,
    })

df_productos = pd.DataFrame(productos)
df_productos
```

```python
lista_lecciones = soup.find_all('div', class_='lesson-description')
```

```python
# 3. Extraer los campos de cada lección
titulo = []
autores = []
enlaces = []
resumenes = []
fechas = []
topicos=[]
actividades=[]
dificultades=[]


for leccion in lista_lecciones:
    titulo.append(leccion.h2.get_text(strip=True))
    autores.append(leccion.h3.get_text(strip=True))
    enlace = leccion.a.get('href')
    enlaces.append(f'https://programminghistorian.org/es/lecciones/{enlace}')
    resumenes.append(leccion.p.get_text(strip=True))
    fechas.append(leccion.find('span', class_='date').get_text())

lecciones_encontradas = {'Titulo':titulo, 'autores': autores, 'Enlaces': enlaces, 'resumenes': resumenes, 'fechas': fechas}
```
